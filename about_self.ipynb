```python
## what is self

class Person():
    def __init__(self, name):
        self.name = name

    def greet(self): ## here , self and p refer to the same thing 
        return f"Hello,{self.name}"



p = Person('Reuven')
p.greet()
```

```python
## (1) self is a  parameter in each method. It is a local variable!
## (2) self is a convention ---not a reserved word.
## (3) every method's first parameter must be self,
##  and it refers to the instance we are running on now.
## (4) reduces ambiguity, makes scoping (LEGB) easier to implement and understand
## (5) when python rewrites our method call, the instance becomes the first arguement
#  
## p.greet() --> Person.greet(p)(more general would be type(p).greet(p))

```
