```python
## x = 10 and y = 20

## I want to swap these values. How? why does this work?

x = 10
y = 20
temp = x
x = y
y = temp

print(f"x = {x}, and y = {y}")
```

```python
x = 10
y = 20
x, y = y, x

print(f"x = {x}, and y = {y}")
```
