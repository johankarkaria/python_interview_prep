```python
## how can i change that if they have same name , they are the same instance object

class Person():
    def __init__(self, name):
        self.name = name

    def __eq__(self, other):
        if not hasattr(other, 'name'):
            return False

        return self.name == other.name


p1 = Person('name')
p2 = Person('name')
```

```python
p1 == p2
```

```python
p1 is p2
```
