```python
class Person():
    def __init__(self, name):
        self.name = name

    def __len__(self):
        return len(self.name)
    
    def __getitem__(self, index):
        return self.name[index]


p = Person('Reuven')
len(p)
```

```python
p[3]   ## p.__getitem__(3)
```

```python
p[2:5]
```
