```python
class Person():
    def __init__(self, name):
        self.name = name

    def __len__(self):
        return len(self.name)
    
    def __getitem__(self, index):
        return self.name[index]


p = Person('Reuven')
len(p)
```

```python
class Person():
    def __init__(self, name):
        self.name = name

    def __len__(self):
        return len(self.name)
    
    def __getitem__(self, index):
        return self.name[index]

    def __add__(self, other):
        return Person(self.name + other.name)


p1 = Person('nam1')
p2 = Person('name2')

```

```python
class Person():
    def __init__(self, name):
        self.name = name

    def __len__(self):
        return len(self.name)
    
    def __getitem__(self, index):
        return self.name[index]

    def __add__(self, other):
        if hasattr(other, 'name'):
            return Person(self.name + other.name)
        else:
            return Person(self.name + str(other))


p1 = Person('nam1')
p2 = Person('name2')


p1 + 'abcd'   # p1.__add__('abcd')
```

```python
class Person():
    def __init__(self, name):
        self.name = name

    def __len__(self):
        return len(self.name)
    
    def __getitem__(self, index):
        return self.name[index]

    def __add__(self, other):
        if hasattr(other, 'name'):
            return Person(self.name + other.name)
        else:
            return Person(self.name + str(other))

    def __radd__(self, other):
        if hasattr(other, 'name'):
            return Person(other.name + self.name)
        else:
            return Person(str(other) + self.name)

    def __repr__(self):
        return f'Person with {vars(self)}'


p1 = Person('nam1')
p2 = Person('name2')
```
