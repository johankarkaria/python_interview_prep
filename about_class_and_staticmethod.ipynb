```python
class Person():
    def __init__(self, name):
        self.name = name

    def greet(self):
        return f"Hello, {self.name}"
        
    @staticmethod
    def hello():
        return "Hello"


p = Person('Reuven')
p.greet()
```

```python
p = Person('Reuven')
Person.hello()
```

```python
p.hello()
```

```python
class Person():
    def __init__(self, name):
        self.name = name

    def greet(self):  ## receives the instance of the Person
        return f"Hello, {self.name}"
        
    @staticmethod
    def hello():  ## don't receive the any class/instance bydefault
        return "Hello"
    

    @classmethod ## receives the CLASS Person
    def Goodbye():
        return 'GoodBye'


p = Person('Reuven')
p.greet()
```

```python
p.hello()
```

```python
Person.hello()
```

```python
Person.Goodbye()
```

```python
class Person():
    def __init__(self, name):
        self.name = name

    def greet(self):  ## receives the instance of the Person
        return f"Hello, {self.name}"
        
    @staticmethod
    def hello():  ## don't receive the any class/instance bydefault
        return f"Hello {self.name}"
    

    @classmethod ## receives the CLASS Person
    def Goodbye(cls):
        return f'GoodBye, {self.name}'


p = Person('Reuven')
p.greet()
```

```python
p.hello()
```

```python
Person.hello()
```

```python
p.Goodbye()
```
