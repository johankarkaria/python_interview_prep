```python
def highest_square_under(max_num):
    if max_num == 0:
        return 0,0
    if max_num == 1:
        return 1,1
    highest_square = 0
    
    for i in range(max_num):
        if i > 2:
            square = i ** 2
            if square > max_num:
                break
            highest_square = square
    return i-1, highest_square



highest_square_under(500)
```
