```python
## Write a function that prints all the names of the files in a directory
## where (string provided as an argument)
## - also print all filenames in subdirectories
## - you can not use os.walk
## - you can ignore errors
```

```python
import os

def print_filenames(dirname):
    all_filenames = os.listdir(dirname)

    for one_filename in all_filenames:
        print(one_filename)

print_filenames('/etc/')
```

```python
import glob
from os.path import join

def print_filenames(dirname):
    all_filenames = glob.glob(join(dirname, '*'))

    for one_filename in all_filenames:
        print(one_filename)

print_filenames('/etc/')
```

```python
import glob
from os.path import join, isdir

def print_filenames(dirname):
    all_filenames = glob.glob(join(dirname, '*'))

    for one_filename in all_filenames:
        if isdir(one_filename):
            print(f'**{one_filename}')
            print(glob.glob(join(one_filename, '*')))
        else:
            print(one_filename)

print_filenames('/etc/')
```

```python
import glob
from os.path import join, isdir

def print_filenames(dirname):
    all_filenames = glob.glob(join(dirname, '*'))

    for one_filename in all_filenames:
        if isdir(one_filename):
            print(f'**{one_filename}')
            all_filenames += glob.glob(join(one_filename, '*'))
        else:
            print(one_filename)

print_filenames('/etc/')
            
```

```python
d = {'a':1, 'b':2, 'c':3}

for key, value in d.items():
    print(f"{key}:{value}")
    if len(key) < 4:
        d[key*2] = value*2      ## a:1 --> aa:2
```

```python
d = {'a':1, 'b':2, 'c':3}

for key, value in list(d.items()):
    print(f"{key}:{value}")
    if len(key) < 4:
        d[key*2] = value*2     ## a:1 --> aa:2

print(d)
```

```python
new_d = {}
for key, value in d.items():
    new_d[key] = value
    if len(key) < 4:
        new_d[key * 2] = value * 2
```
