```python
## what are *args and **kwargs
def hello(name):
    return f"Hello, {name}"

hello("johan")
```

```python
hello("a", "b")
```

```python
## the * indicate that the args will be a tuple and it would contain all the argument 
## that woud be passed to the function

def hello(*args):
    return f"hello {args}"

hello("a","b","c")
```

```python
## we can not have a multiple args in a function 
def hello(*arg1, *arg2):
    return f"hello {arg1, arg2}"
```

```python
## we could have something like this and it would not throw any error
def hello(arg, *args2):
    return f"hello {arg,args2}"
```

```python
hello("a", "b", "c", "d", "e")
```

```python
def hello(a, *args2):
    return f"hello a = {a}, args2 = {args2}"

hello(10, 20, 30, 40, 50, 60, 70, 80, 90)
```

```python
## as you can see it throws an error , that we can not keep b as a positional argument
## after *args to keep the b as parameter and it works fine we must keep it as 
# keyword argument

def hello(a, *args, b):
    return f"hello a = {a}, args = {args}, b = {b}"

hello(10, 20, 30, 40, 50, 60, 70)
```

```python
hello(10, 20, 30, 40, 50, 60, b = 70)
```

```python
hello(10, 20, 30, b = 100)
```

```python
## we could not do that as python don't recognize x as a parameter
hello(10, 20, 30, x = 40)
```

```python
## inorder to run the above function without an error but giving any keyword parameter
## we must us *kwargs
## kwargs only picks up keyword arguments

def hello(a, *args, **kwargs):
    return f"a = {a}, args = {args}, kwargs = {kwargs}"

hello(10, 20,30, 40, 50)
```

```python
hello(10, 20, 30, 40, 50, x=100, y = 200, z = 300)
```
