```python
## write a generator function, read_n, that would take two arguments
## a filename and a number (n).In each iteration, the generator should return n line from the file.
## if there aren't n line remaining, then it should return as many line as possible.



## cautions: we don't get one line of the file because the variable , 
# but because of the iteration protocol

## some files not may not be read only one line at a time, they may be read 5 line at a time,
#  2 line at a time.


for one_line in open('/etc/passwd'):
    print(one_line)

```

```python
def read_n(filename, n):
    for one_line in open(filename):
        yield one_line


for one_chunk in read_n('/etc/passwd', 5):
    print(one_chunk)
```

```python
def read_n(filename, n):
    with open(filename) as f:
        while True:
            sentences = ''
            for _ in range(n):
                one_line = f.readline()
                if not one_line:
                    break
                sentences += one_line + '\n'
            if not sentences:
                break
            yield sentences
            
    

for one_chunk in read_n('/etc/passwd', 5):
    print(one_chunk, end='')
```
