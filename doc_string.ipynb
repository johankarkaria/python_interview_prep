```python
def hello(first, last=''):
    '''
    This is the world's best function!
    It's so great, it gets a two-line doc string!
    '''
    return f"Hello, {first} {last}"

hello("Reuven")
```

```python
help(hello)
```

```python
def hello(first, last=''):
    """
    Expect:one or two string, representing first + lastname
    Modifies:Nothing
    Returns: a string(a friendly greeting using the person's name)
    """
    return f"Hello, {first} {last}"

hello("Reuven")
```

```python
help(hello)
```
