```python
## re-implement 'range' to return an iterator

## range(5) -- iterate from 0 to 4
## range(5, 10) -- iterate from 5 to 9
## range(5, 20, 2) -- iterate from 5-19, by 2s (5-7-9-11-13-15-17-19)
```

```python
def myrange(first, second=None, step=1):
    if second is None:
        current = 0
        maxnum = first
    else:
        current = first
        maxnum = second

    while current < maxnum:
        yield current
        current += step


list(myrange(5))
```

```python
list(myrange(5,10))
```

```python
list(myrange(5,20,2))
```

```python
list(myrange(10, step=2))
```

```python
list(range(10, step=2))
```

```python
list(range(10, 2, -2))
```

```python
def myrange(first, second=None, step=1):
    if second is None:
        current = 0
        maxnum = first
    else:
        current = first
        maxnum = second

    if step > 0:
        while current < maxnum:
            yield current
            current += step
    else:
        while current > maxnum:
            yield current
            current += step

list(myrange(5))
```

```python
list(myrange(20, 2, -3))
```
