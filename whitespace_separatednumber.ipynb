```python
## get input from the  users --- white space-separated-integers
## in a string. Get a list of integers (not a string or list of strings),
## sorted from smallest to largest based on this input

s = '10 20 15 5 30 27'
## from this we want to get 

## [5 10 15 20 27 30]
```

```python
s.split()  ## no argument == split on all whitespace, any number of chars
```

```python
sorted(s.split())
```

```python
sorted([int(one_number) 
for one_number in s.split()])
```

```python
## this also works

sorted(int(one_number) 
for one_number in s.split())
```

```python
s = '10 20 test abcd 15 5 xyz 30 27'
sorted(int(one_number) 
for one_number in s.split() 
if one_number.isdigit())
```
