```python
## what is a closure ?? (and give an example)

## when i use 'def' , I both create a new function object
## and define a variable
## when I define a variable inside a function, I'm creating a local variable
## - a function can return any Python object, without exception 


def foo(): ## outer function
    def bar(): ## inner function
        print(f"I am in bar!")
    return bar

```

```python
def foo(x): 
    def bar(y):
        print(f"I am in bar, x = {x} and y = {y}!")
    return bar
```

```python
b1 = foo(10) ## bar is a closure -- has access to foo's local variable!
b2 = foo(20) 
```

```python
b1(3)
```

```python
b2(4)
```
