```python
def foo(x): 
    def bar(y):
        print(f"I am in bar, x = {x} and y = {y}!")
    return bar
```

```python
b1 = foo(10) ## bar is a closure -- has access to foo's local variable!
b2 = foo(20) 
```

```python
def count_calls(func):
    count = 0
    def call_func(*args, **kwargs):
        nonlocal count ## this means: assign to count_calls's count
        count += 1
        print(f"Calling {func}, time, {count}")
        return func(*args, **kwargs)
    return call_func


@count_calls
def hello(name):
    return f"Hello, {name}"

## hello = count_calls(hello)


hello('world')
```

```python
c1 = count_calls(hello)
```
