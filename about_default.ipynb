```python
## we should not be storing the variable's value in defaults that is bad
## instead we can do something like this


def add_one(x=None):
    if x is None: ## assigning during runtime
        x = []
    x.append(1)
    return x
    
```

```python
add_one()
```

```python
add_one()
```
