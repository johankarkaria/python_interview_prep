```python
def read_n(filename, n):
    with open(filename) as f:
        while True:
            one_chunk =  ''.join(f.readline() for i in range(n))
            if one_chunk:
                yield one_chunk
            else:
                break
        
gen = read_n('virat_kohli.txt',3)
next(gen)
```
