```python
## write a decorator that ensure all of the arguments passed to a function
## will be  strings

## only_string_args will be a function 
## that takes a function arguments
## returns a function as a results

def only_string_args(func):
    def wrapper(*args, **kwargs):
        return func(*[one_arg
                        for one_arg in args
                        if isinstance(one_arg, str)], **kwargs)
    return wrapper




@only_string_args
def hello(name):
    return f"Hello, {name}"

@only_string_args
def goodbye(name):
    return f"Goodbye, {name}"

@only_string_args
def word_length(*words):
    return {one_word: len(one_word)
            for one_word in words}


print(hello('world'))
```

```python
print(goodbye('world'))
```

```python
print(hello(12345,'abc'))
```

```python
word_length('a', 'bcd', 'efgh', 5)
```

```python
def only_typed_args(the_type):
    def middle(func):
        def wrapper(*args, **kwargs):
            return func(*[one_arg
                            for one_arg in args
                            if isinstance(one_arg, the_type)], **kwargs)
        return wrapper
    return middle




@only_typed_args(str)
def hello(name):
    return f"Hello, {name}"

@only_typed_args(str)
def goodbye(name):
    return f"Goodbye, {name}"

@only_typed_args(str)
def word_length(*words):
    return {one_word: len(one_word)
            for one_word in words}

@only_typed_args(int)
def div(a, b):
    return a/b

print(hello('world'))
```
