```python
## break vs continue

for one_character in 'abcde':
    print(one_character)
```

```python
print("** Start **")
for one_character in 'abcde':
    if one_character == 'c':
        break ## stop the loop right now.
    print(one_character)


print("** End **")
```

```python
x = 5
while x > 0:
    print(x)
    x -= 1
```

```python
x = 5
while x > 0:
    if x == 2:
        break
    print(x)
    x -= 1
```

```python
while True:
    name = input("Enter your name: ").strip()

    if not name: ## meaning: did we get a blank string
        break
    print(f"hello, {name}")
```

```python
print("** Start **")
for one_character in 'abcde':
    if one_character == 'c':
        continue ## stop the current iteration right now
    print(one_character)


print("** End **")
```

```python
for one_line in open('/etc/passwd'):
    print(one_line)
```

```python
for one_line in open('/etc/passwd'):
    if one_line.startswith('#'):
        continue  ## go on to the next line /iteration

    print(one_line.split(":")[0])
```
