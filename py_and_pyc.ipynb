```python
## Python tries to be smart about importing modules

## if there is a .py and NO .pyc , Python creates the .pyc
## and then loads the  .pyc

## if there is a .pyc but no .py, Python loads .pyc


## if there is a .py and AND a .pyc, and the .pyc is NEWER,
## then it (the .pyc is loaded)


## if there is a .py AND a .pyc and the .py is NEWER ,
## then the python removes the .pyc and recompile the module,
## creating NEW .pyc, which is then loaded 
```
