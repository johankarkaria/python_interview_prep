```python
s = 'abcdef'
look_for ='d'

found_it = False
for one_letter in s:
    if look_for == one_letter:
        found_it = True
        break
        
print(f"Found {look_for} in {s}: {found_it}")
```

```python
s = 'abcdef'
look_for ='d'


for one_letter in s:
    if look_for == one_letter:
        print(f"Yay!  Found {look_for}")
        break
else: ## runs only if the loop got to its natural end -- no break
    print(f"Did *Not* find {look_for}")


## else == 'reached end of the bock without encountering a break'
```
