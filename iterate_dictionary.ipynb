```python
## how can you iterate over a dictionary ??

d = {'a':1, 'b':2, 'c':3, 'd':4}

for one_item in d:
    print(one_item)
```

```python
for one_item in d.keys():
    print(one_item)
```

```python
d.keys()
```

```python
type(d.keys())
```

```python
for one_key in d:
    print(f"{one_key}:{d[one_key]}")
```

```python
d.items()
```

```python
for t in d.items():
    print(f"{t[0]}:{t[1]}")
```

```python
for key, value in d.items():
    print(f"{key}:{value}")
```
