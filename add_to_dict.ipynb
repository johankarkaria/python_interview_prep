```python
## adding to dict (how do we add to a dictionary)

d = {'a':1, 'b':2, 'c':3}

#c/4
#d/5


d['c']=4 ## because 'c' already exists, we'll replace 3 --> 4 
```

```python
d
```

```python
## because 'd' is new key and its value is added
d['d']=5
d
```

```python
## pre-3.6 python : order of key, value pair was not defined.

## starting in 3.6:order of key, value pair is chronological.


d['q']=100
d['j']=200
d['x']=300
d['l']=400

d
```

```python
## remember that dictionary key must be hashable
mylist = [10, 20, 30, 40]
d[mylist] = 1000
```

```python
hash(mylist)
```
